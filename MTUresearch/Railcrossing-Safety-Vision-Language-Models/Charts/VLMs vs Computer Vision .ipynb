```python
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# DATA
# ============================================================

tasks = [
    "Multi-class Classification",
    "Vehicle Type Classification",
    "Vehicle Intensity Classification"
]

# ResNet-50 Macro F1 baseline
resnet = np.array([
    96.0,
    70.0,
    49.0
])

# Zero-shot VLM Macro F1 scores
gemini = np.array([
    100.0,
    79.0,
    70.5
])

gpt = np.array([
    97.74,
    76.6,
    90.3
])

claude = np.array([
    98.51,
    79.7,
    90.7
])

vlm_data = [
    ("Gemini-3.5-Flash", gemini),
    ("GPT-4o-mini", gpt),
    ("Claude-Sonnet-5", claude)
]

# ============================================================
# COLORS
# ============================================================

vlm_colors = {
    "Gemini-3.5-Flash": "#BCBD22",
    "GPT-4o-mini": "#17BECF",
    "Claude-Sonnet-5": "#1F77B4"
}

# ============================================================
# STYLE
# ============================================================

plt.rcParams["font.family"] = "Times New Roman"
plt.rcParams["font.size"] = 14

fig, ax = plt.subplots(figsize=(10, 6))

x = np.arange(len(tasks))
offsets = [-0.18, 0.00, 0.18]

# ============================================================
# DRAW BASELINE-TO-VLM COMPARISONS
# ============================================================

for offset, (model_name, values) in zip(offsets, vlm_data):

    positions = x + offset
    model_color = vlm_colors[model_name]

    for i, xpos in enumerate(positions):

        # IMPORTANT: ResNet-50 is the baseline
        baseline = resnet[i]
        vlm_value = values[i]

        # Percentage-point improvement over ResNet-50
        difference = vlm_value - baseline

        # Connecting line
        ax.vlines(
            xpos,
            baseline,
            vlm_value,
            color=model_color,
            linewidth=1.6,
            alpha=0.85,
            zorder=2
        )

        # ResNet-50 baseline marker
        ax.scatter(
            xpos,
            baseline,
            marker="_",
            s=180,
            linewidth=2.2,
            color="black",
            zorder=3
        )

        # VLM circular marker
        ax.scatter(
            xpos,
            vlm_value,
            marker="o",
            s=85,
            color=model_color,
            zorder=4
        )

        # ----------------------------------------------------
        # VALUE + PERCENTAGE-POINT DIFFERENCE
        # ----------------------------------------------------

        if vlm_value >= 96:
            label_y = vlm_value + 1.2
        else:
            label_y = vlm_value + 1.0

        ax.text(
            xpos,
            label_y,
            f"{vlm_value:.1f}\n(+{difference:.1f})",
            ha="center",
            va="bottom",
            fontweight="bold",
            fontsize=12,
            color="black",
            clip_on=False
        )

# ============================================================
# RESNET-50 BASELINE LABELS
# ============================================================

baseline_label_x = [
    x[0] - 0.34,
    x[1] - 0.34,
    x[2] - 0.34
]

for i, value in enumerate(resnet):

    ax.text(
        baseline_label_x[i],
        value,
        f"ResNet-50 = {value:.1f}",
        ha="right",
        va="center",
        fontsize=9,
        fontweight = 'bold',
        fontstyle="italic",
        color="black"
    )

# ============================================================
# AXES
# ============================================================

ax.set_xticks(x)

ax.set_xticklabels(
    tasks,
    fontsize=13
)

ax.set_ylabel(
    "Macro F1-Score (%)",
    fontweight="bold",
    fontsize=14
)

# Same scale for both figures
ax.set_ylim(40, 108)

# Extra space on left for baseline labels
ax.set_xlim(-0.70, 2.32)

ax.tick_params(
    axis="y",
    labelsize=12
)

# ============================================================
# GRID AND SPINES
# ============================================================

ax.grid(
    axis="y",
    linestyle=":",
    linewidth=0.8,
    alpha=0.35
)

ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# ============================================================
# LEGEND
# ============================================================

for model_name, _ in vlm_data:

    ax.scatter(
        [],
        [],
        marker="o",
        s=70,
        color=vlm_colors[model_name],
        label=model_name
    )

ax.legend(
    frameon=False,
    loc="lower left",
    fontsize=12
)

# ============================================================
# LAYOUT
# ============================================================

plt.tight_layout(pad=1.5)

# ============================================================
# OPTIONAL SAVE
# ============================================================

# plt.savefig(
#     "VLM_MacroF1_vs_ResNet50.png",
#     dpi=600,
#     bbox_inches="tight",
#     pad_inches=0.1
# )

plt.show()
```
