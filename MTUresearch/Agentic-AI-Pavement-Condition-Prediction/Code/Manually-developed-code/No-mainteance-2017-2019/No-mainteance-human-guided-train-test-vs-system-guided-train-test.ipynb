```python
import pandas as pd

# Load both files
agent = pd.read_csv("C:\\Users\\tadnan\\Downloads\\Agentic-AI-Pavement-Condition_Predictions\\Outputs\\GPT-5.5-No-maintenance\\run-3-GPT-5.5-No-maintenance-output\\features\\train_valid_test_indices.csv")
human = pd.read_csv("C:\\Users\\tadnan\\Downloads\\Agentic-AI-Pavement-Condition_Predictions\\Human-guided-code\\No-mainteance-2017-2019\\human_guided_train_valid_test_indices.csv")

# Standardize split names
agent["split"] = agent["split"].str.lower().str.strip()
human["split"] = human["split"].str.lower().str.strip()

# Extract training indices
agent_train = set(
    agent.loc[agent["split"] == "train", "row_position_after_target_drop"]
)

human_train = set(
    human.loc[human["split"] == "train", "row_position_after_target_drop"]
)

# Compare
common_train = agent_train.intersection(human_train)
only_agent = agent_train - human_train
only_human = human_train - agent_train

print(f"Agent train samples : {len(agent_train)}")
print(f"Human train samples : {len(human_train)}")
print(f"Common train samples: {len(common_train)}")
print(f"Only in Agent train : {len(only_agent)}")
print(f"Only in Human train : {len(only_human)}")

agreement = len(common_train) / len(agent_train) * 100
print(f"\\nTraining-set agreement: {agreement:.2f}%")
```
