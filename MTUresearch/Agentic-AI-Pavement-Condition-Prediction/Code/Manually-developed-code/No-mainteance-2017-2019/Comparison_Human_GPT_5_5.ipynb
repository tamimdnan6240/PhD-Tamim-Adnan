```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Models
# -----------------------------
models = [
    "Random Forest",
    "Gradient Boosting",
    "HistGradientBoosting",
    "Extra Trees",
    "Elastic Net"
]

x = np.arange(len(models))
width = 0.35

# -----------------------------
# Agentic AI Workflow results
# -----------------------------
agentic_no_maintenance_r2 = [
    0.884795260768644,
    0.887020570388342,
    0.894348087748373,
    0.892241116059165,
    0.884238172141449
]

agentic_no_maintenance_mae = [
    11.0484505463788,
    10.9682542557657,
    10.7493467479813,
    10.8336821656138,
    11.115337902266
]

agentic_no_maintenance_rmse = [
    20.6219345177235,
    20.4217953513892,
    19.7484453210677,
    19.944390702738,
    20.6717345107239
]

agentic_no_maintenance_mape = [
    8.85289264528496,
    9.00965072458041,
    8.73361407199286,
    8.85311375507283,
    9.16032272109068
]

# -----------------------------
# Manually Developed Workflow results
# -----------------------------
human_no_maintenance_r2 = [
    0.9102,
    0.9141,
    0.9106,
    0.9095,
    0.9084
]

human_no_maintenance_mae = [
    10.2074,
    9.8801,
    10.0022,
    10.3826,
    10.4261
]

human_no_maintenance_rmse = [
    17.2513,
    16.8711,
    17.2169,
    17.3175,
    17.4216
]

human_no_maintenance_mape = [
    8.8451,
    8.5446,
    8.6452,
    9.0238,
    9.1802
]

# -----------------------------
# Create figure
# -----------------------------
fig, axes = plt.subplots(2, 2, figsize=(15, 9))

# -----------------------------
# Annotation function
# -----------------------------
def annotate_bars(ax, bars, decimals=2, suffix=""):
    for bar in bars:
        height = bar.get_height()
        ax.annotate(
            f"{height:.{decimals}f}{suffix}",
            xy=(bar.get_x() + bar.get_width() / 2, height),
            xytext=(0, 3),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=8
        )

# -----------------------------
# (a) R²
# -----------------------------
bars1 = axes[0, 0].bar(
    x - width / 2,
    agentic_no_maintenance_r2,
    width,
    label="Agentic AI Workflow"
)

bars2 = axes[0, 0].bar(
    x + width / 2,
    human_no_maintenance_r2,
    width,
    label="Manually Developed Workflow"
)

annotate_bars(axes[0, 0], bars1, decimals=3)
annotate_bars(axes[0, 0], bars2, decimals=3)

axes[0, 0].set_title("(a) R² – No-maintenance")
axes[0, 0].set_ylabel("R²")
axes[0, 0].set_ylim(0.80, 0.95)
axes[0, 0].grid(axis="y", alpha=0.3)

# -----------------------------
# (b) MAE
# -----------------------------
bars3 = axes[0, 1].bar(
    x - width / 2,
    agentic_no_maintenance_mae,
    width,
    label="Agentic AI Workflow"
)

bars4 = axes[0, 1].bar(
    x + width / 2,
    human_no_maintenance_mae,
    width,
    label="Manually Developed Workflow"
)

annotate_bars(axes[0, 1], bars3)
annotate_bars(axes[0, 1], bars4)

axes[0, 1].set_title("(b) MAE – No-maintenance")
axes[0, 1].set_ylabel("MAE")
axes[0, 1].set_ylim(0, 15)
axes[0, 1].grid(axis="y", alpha=0.3)

# -----------------------------
# (c) RMSE
# -----------------------------
bars5 = axes[1, 0].bar(
    x - width / 2,
    agentic_no_maintenance_rmse,
    width,
    label="Agentic AI Workflow"
)

bars6 = axes[1, 0].bar(
    x + width / 2,
    human_no_maintenance_rmse,
    width,
    label="Manually Developed Workflow"
)

annotate_bars(axes[1, 0], bars5)
annotate_bars(axes[1, 0], bars6)

axes[1, 0].set_title("(c) RMSE – No-maintenance")
axes[1, 0].set_ylabel("RMSE")
axes[1, 0].set_ylim(0, 24)
axes[1, 0].grid(axis="y", alpha=0.3)

# -----------------------------
# (d) MAPE
# -----------------------------
bars7 = axes[1, 1].bar(
    x - width / 2,
    agentic_no_maintenance_mape,
    width,
    label="Agentic AI Workflow"
)

bars8 = axes[1, 1].bar(
    x + width / 2,
    human_no_maintenance_mape,
    width,
    label="Manually Developed Workflow"
)

annotate_bars(axes[1, 1], bars7, suffix="%")
annotate_bars(axes[1, 1], bars8, suffix="%")

axes[1, 1].set_title("(d) MAPE – No-maintenance")
axes[1, 1].set_ylabel("MAPE (%)")
axes[1, 1].set_ylim(0, 14)
axes[1, 1].grid(axis="y", alpha=0.3)

# -----------------------------
# Common x-axis formatting
# -----------------------------
for ax in axes.flat:
    ax.set_xticks(x)
    ax.set_xticklabels(models, rotation=20, ha="right")

# -----------------------------
# Common legend
# -----------------------------
handles, labels = axes[0, 0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    loc="lower center",
    ncol=2,
    fontsize=11,
    frameon=False
)

# -----------------------------
# Layout
# -----------------------------
plt.tight_layout(rect=[0, 0.08, 1, 1])

# -----------------------------
# Save figure
# -----------------------------
plt.savefig(
    "gpt-5.5_agentic_vs_manual_workflow_no_maintenance_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
