```python
original_counts = df_orignal['Maintenance Intervention'].value_counts()
original_counts
```

```python
# Load your sample dataset
df_sample = pd.read_csv('/content/drive/MyDrive/ LLM application for Maintenance Decision Making/Data/sample_df_500.csv')
```

```python
sample_counts = df_sample['Maintenance Intervention'].value_counts()
sample_counts
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Percentages
original_pct = original_counts / original_counts.sum() * 100
sample_pct = sample_counts / sample_counts.sum() * 100

# Create labels with original counts
labels = [
    f"{cat}\n{original_counts[cat]:,} → {sample_counts[cat]:,}"
    for cat in original_counts.index
]

comparison_df = pd.DataFrame({
    'Original (%)': original_pct.values,
    'Sample (%)': sample_pct.values
}, index=labels)

# Plot
ax = comparison_df.plot(
    kind='bar',
    figsize=(12,6),
    width=0.8
)

# Add percentage annotations
for container in ax.containers:
    ax.bar_label(
        container,
        fmt='%.1f%%',
        padding=3,
        fontsize=9
    )

plt.ylabel('Percentage (%)')
plt.xlabel('Maintenance Intervention')
plt.title(
    'Distribution of Maintenance Intervention Categories\n'
    'Original Dataset (n=23,632) vs Sample Dataset (n=500)'
)
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()
```
